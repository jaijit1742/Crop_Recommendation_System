# Duplicate Values

```python
import pandas as pd
df = pd.read_csv('Dataset/Raw/Crop_recommendation.csv')
print('First few rows of the raw dataset:')
print(df.head())
```

```python
print('Duplicate records:', df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print('Records after removing duplicates:', len(df))
```

```python
print('Duplicate records after removal:', df.duplicated().sum())
df.head()
```
